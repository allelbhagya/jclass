```python
import pandas as pd

df = pd.read_csv("../data/data_main.csv")
```

```python
print(df.shape)
print(df.head())
print(df.columns)
```

```python
print(df["label"].value_counts())
```

```python
text_columns = [
    "job_title",
    "job_description",
    "experience_requirement",
    "seniority_terms",
    "leadership_responsibilities"
]

df["text"] = df[text_columns].fillna("").agg(" ".join, axis=1)

print(df["text"].iloc[0])
```

```python
print(df[["text", "label"]].isna().sum())
```

```python
print(df["label"].value_counts())
```

```python
print(
    df[
        [
            "job_title",
            "min_years_experience",
            "max_years_experience",
            "experience_required_or_preferred",
            "internship_indicator",
            "new_grad_indicator",
            "fresher_indicator",
            "label",
        ]
    ].to_string(index=False)
)
```

```python
print(
    df.groupby("label")[
        ["min_years_experience", "max_years_experience"]
    ].agg(["min", "max", "mean"])
)
```
